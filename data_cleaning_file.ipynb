```python
# clean Customer name
for val in df["Name"]:
    name=val
    name=" ".join(val.split()).title()#extra space remove+proper capitalization
    val=name
    print(val)
```

```python
#clean Age
for val in df["Age"]:
    if val>80:print(val)
#display Age
df["Age"]
```

```python
#clean city name and proper
for val in df["City"]:
    name=val
    name=" ".join(val.split()).title()
    val=name
    print(val)
```

```python
# validate Email


df["Email"]=df["Email"].str.replace(".","",regex=False)

df["Email"]=df["Email"].str.split("@").str[0]
df["Email"]=df["Email"]+"@gmail.com"
print(df["Email"])
```

```python
for val in df["Phone"]:
    if val!=10:
        print("invalid phone number:",val)
```

```python
#remove phone
print(df.dropna(subset=["Phone"]))
print(df.isnull().sum())

#fill na phone
df["Phone"]=df["Phone"].fillna("not available")
print(df.isnull().sum())

# replace (+91,91+,space,(,),_)remove
print(df.dtypes)
df["Phone"]=df["Phone"].str.strip()
df["Phone"]=df["Phone"].str.replace(" ","",regex=False)
df["Phone"]=df["Phone"].str.replace("-","",regex=False)
df["Phone"]=df["Phone"].str.replace("(","",regex=False)
df["Phone"]=df["Phone"].str.replace(")","",regex=False)
df["Phone"]=df["Phone"].str.replace("91+","",regex=False)
df["Phone"]=df["Phone"].str.replace("+91","",regex=False)
df["Phone"]=df["Phone"].str.replace(".0","",regex=False)
print(df["Phone"])
```

```python
#missing value and missing value ko remove
print(df["SignupDate"].isnull().sum())
print(df.dropna(subset=["SignupDate"]))
```

```python
#duplicate value
df.duplicated(subset=["SignupDate"])
df.drop_duplicates(subset=["SignupDate"])
```

```python
# date correct formate
for val in df["SignupDate"]:
    val = pd.to_datetime(val)
    val = val.strftime("%d-%m-%Y")
    print(val)
```

```python
df.to_csv("cleaned_customer_data.csv", index=False)

print("Cleaned CSV file saved successfully!")
```
